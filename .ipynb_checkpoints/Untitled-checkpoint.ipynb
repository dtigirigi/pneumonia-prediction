```python
import os
import cv2
import numpy as np
import pandas as pd
from PIL import Image
import mahotas as mh

# Folder paths
BASE = os.path.join("Dataset", "Chest Xray Images Full")

TRAIN_DIR = os.path.join(BASE, "Train")
TEST_DIR  = os.path.join(BASE, "Test")
VAL_DIR   = os.path.join(BASE, "Val")

IMG_SIZE  = (224, 224)

# Load Metadata
def load_metadata(root):
    rows = []
    for label in ["NORMAL", "PNEUMONIA"]:
        folder = os.path.join(root, label)
        for f in os.listdir(folder):
            if f .lower().endswith((".jpg", ".jpeg", ".png")):
                rows.append([f, label, os.path.join(folder, f)])
    return pd.DataFrame(rows, columns=["Image_ID", "Label", "File_Path"])

train_meta = load_metadata(TRAIN_DIR)
test_meta  = load_metadata(TEST_DIR)
val_meta   = load_metadata(VAL_DIR)

# Check for corrupted images
def find_bad_images(df):
    bad = []
    for path in df["File_Path"]:
        try:
            Image.open(path)
        except Exception as e:
            bad.append((path,str(e)))
    return bad

bad_train = find_bad_images(train_meta)
bad_test  = find_bad_images(test_meta)
bad_val   = find_bad_images(val_meta)

print("Bad TRAIN images:", bad_train)
print("Bad TEST images:", bad_test)
print("Bad VAL images:", bad_val)


# Remove corrupted images
def remove_bad(df, bad_list):
    bad_paths = [b[0] for b in bad_list]
    return df[~df["File_Path"].isin(bad_paths)]

train_meta = remove_bad(train_meta, bad_train)
test_meta  = remove_bad(test_meta, bad_test)
val_meta   = remove_bad(val_meta, bad_val)

# Feature Extraction Functions
def extract_intensity_features(img):
    return {
        "Mean Intensity": np.mean(img),
        "Std Intensity": np.std(img),
        "Min Intensity": np.min(img),
        "Max Intensity": np.max(img)
    }

def extract_haralick_features(img):
    # Mahotas expects grayscale uint8
    img_uint8 = img.astype(np.uint8)
    har = mh.features.haralick(img_uint8)
    har_mean = har.mean(axis=0)
    return {f"Haralick_{i+1}": har_mean[i] for i in range(len(har_mean))}

# Extract Features for ALL images (Train, Test, Val)
def extract_features(df):
    feature_rows = []

    for _, row in df.iterrows():
        path =row["File_Path"]
        label = row["Label"]

        img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
        img = cv2.resize(img, IMG_SIZE)

        intensity = extract_intensity_features(img)
        haralick = extract_haralick_features(img)

        all_features = {
            "Image_ID": row["Image_ID"],
            "Label": label,
            **intensity,
            **haralick
        }

        feature_rows.append(all_features)

    return pd.DataFrame(feature_rows)

train_features = extract_features(train_meta)
test_features = extract_features(test_meta)
val_features = extract_features(val_meta)

# Build Master Dataset for Train
master_df = train_features.copy()

# Binary encode label
master_df["Label_Binary"] = (master_df["Label"] == "PNEUMONIA").astype(int) 

print("\nMaster DF Ready:")
print(master_df.head())
print("\nShape:", master_df.shape)
        
```

## PHASE 2 - Exploratory Data Analysis (EDA)

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")

# Count Normal vs Pneumonia and Check class imbalance
plt.figure(figsize=(6,4))
sns.countplot(data=master_df, x="Label", palette="viridis")
plt.title("Distribution of Normal vs Pneumonia Cases")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()

print(master_df["Label"].value_counts())
print("\nClass Imbalance (%):")
print(master_df["Label"].value_counts(normalize=True) * 100)

# Feature Analysis
# Summary Statistics (Normal vs Pneumonia vs Overall)
num_features = ["Mean_Intensity", "Std_Intensity", "Min_Intensity", "Max_Intensity"]

summary_table = pd.DataFrame()

for feat in num_features:
    summary_table.loc[feat, "Normal_Mean_SD"] = (
        f"{master_df[master_df.Label=='NORMAL'][feat].mean():.2f} "
        f"({master_df[master_df.Label=='NORMAL'][feat].std():.2f})"
    )
    summary_table.loc[feat, "Pneumonia_Mean_SD"] = (
        f"{master_df[master_df.Label=='PNEUMONIA'][feat].mean():.2f} "
        f"({master_df[master_df.Label=='PNEUMONIA'][feat].std():.2f})"
    )
    summary_table.loc[feat, "Overall_Mean_SD"] = (
        f"{master_df[feat].mean():.2f} ({master_df[feat].std():.2f})"
    )

summary_table

# Boxplots by Class
plt.figure(figsize=(12,8))
for i, feat in enumerate(num_features):
    plt.subplot(2,2,i+1)
    sns.boxplot(data=master_df, x="Label", y=feat, palette="coolwarm")
plt.title(f"{feat} by Class")
plt.tight_layout()
plt.show()

# Create Intensity Categories
master_df["Intensity_Category"] = pd.qcut(
    master_df["Mean_Intensity"],
    q=3,
    labels=["Low Intensity", "Medium Intensity", "High Intensity"]
)

# Pneumonia Rate by Intensity Category
table2 = master_df.groupby("Intensity_Category").agg(
    N=("Label","count"),
    Pneumonia_Cases=("Label_Binary","sum")
)

table2["Pneumonia_Rate_%"] = (table2["Pneumonia_Cases"] / table2["N"]) * 100
table2

# Bar Chart for Categorical Features
plt.figure(figsize=(6,4))
sns.barplots(data=table2.reset_index(),
             x="Intensity_Category",
             y="Pneumonia_Rate_%",
             palette="magma")
plt.title("Pneumonia Rate by Intensity Category")
plt.ylabel("Pneumonia Rate (%)")
plt.show()

# Multivariate analysis - interaction of multiple features
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=master_df,
    x="Mean_Intensity",
    y="Haralick_1",
    hue="Label",
    palette="coolwarm",
    alpha=0.7
)
plt.title("Mean Intensity vs Haralick_1 (Texture) by Class")
plt.show()

# Correlation Analysis - Indentify redundant features
plt.figure(figsize=(14,10))
corr = master_df[num_features + [f"Haralick_{i+1}" for i in range (13)]].corr()
sns.heatmap(corr, cmap="coolwarm", annot=False)
plt.title("Correlation Matrix of Derived Features")
plt.show()

     
```
